check if number is even or odd


```python
def is_even(n):
  if n%2==0:
    return "even"
  else:
    return "odd"
is_even(5)
```

Factorial of a number

```python
def factorial(n):
  fact=1
  for i in range(1,n+1):
    fact=fact*i
  print(fact)
factorial(20)
```

Largest Of Three Numbers

```python
def largest(*a):
  return max(a)
largest(0,99,88,66,432,999)
```

Reverse a String

```python
def reverse_string(s):
  return s[::-1]
reverse_string("Pavan")
```

count Vowlels in a string

```python

def count_vowels(k):
  vowels="aeiouAEIOU"
  count=0
  for ch in k:
    if ch in vowels:
      count+=1
  return count
count_vowels("Pavankumar")
```

Check if string is a Palindrome

```python
def is_palindrome(s):
  return s==s[::-1]
is_palindrome("MadaM")
```

Sum of all numbers in a list

```python
def lst_sum(lst):
  total=0
  for i in lst:
    total=total+i
  return total
lst_sum([9,8,9,0,5,])
```

Print multiplication table of number

```python
def table(m):
  for i in range(1,20):
    print(m,"*" ,i,"=",m*1)
table(10)
```

count words in a sentence

```python
def count_word(sentence):
  words=sentence.split()
  return len(words)
count_word("I LOVE MY PARENTS AND MY COUNTRY")
```

Return square of each number in a list

```python
def square_lst(lst):
  squares=[]
  for num in lst:
    squares.append(num**2)
  return squares
square_lst([1,2,3,4])
```

Types of Arguments in Python Functions

Python supports 5 main types of arguments:

  Positional Arguments

  Keyword Arguments

  Default Arguments

  Variable-length Arguments → *args

  Variable-length Keyword Arguments → **kwargs

Positional Arguments
  
  Arguments that are passed in the *same order as parameters in function definition.*

```python
def info(name,age):
  print("Name",name)
  print("Age",age)
info("Pavan",25)
```

```python
def subtract(a,b):
  return a-b
subtract(30,100)
```

```python
def fullname(fname,lname):
  print(fname+" "+lname)
fullname("Pavan","Kumar")
```

```python
def check(num):
  if num%2==0:
    print("Even")
  else:
    print("Odd")
check(20)
```

```python
def area(length,Width):
  return length*Width
print(area(5,3))
```

Variable-length Arguments (*args)

Used when the number of arguments is unknown.


*args  collects all extra positional arguments into a tuple.


```python
def find_sum(*nums):
  return sum(nums)
find_sum(10,20,30,40)
```

NOTE-args stores data as a tuple.

```python
def multiply(*nums):
  product=1
  for i in nums:
    product=product*i
  return product
multiply(2,3,4,5)
```

```python
def count_even(*nums):
    count = 0
    for n in nums:
        if n % 2 == 0:
            count += 1
    print("Even count:", count)

count_even(1, 2, 5, 8, 10)
```

```python
def average(*nums):
    print(sum(nums)/len(nums))

average(10, 20, 30)
```

Keyword Variable-length Arguments → **kwargs


   Used for unknown number of keyword arguments.

   kwargs stores data as a dictionary.


```python
def info(**kwargs):
  print(kwargs)
info(name="Pavan",age=22,city='Banglore')
```

```python
def student_deatails(**details):
  for key,value in details.items():
    print(key,":",value)
student_deatails(name="PavanKumar",Age=25,course="Python Developer")
```

```python
def book(**info):
  for k,v in info.items():
    print(k,":",v)
book(title="Harry Potter", author="Rowling", price=450)
```

```python
def count_kwargs(**data):
  return len(data)
count_kwargs(a=10,b=30,c=90,d=100,f=88)
```
